```python
from pathlib import Path

# Localizar la actividad desde su raíz o desde notebooks.
actual = Path.cwd().resolve()

base = next(
    (
        carpeta
        for carpeta in (actual, *actual.parents)
        if carpeta.name.lower() == "pre_11_vuelos"
    ),
    None,
)

if base is None:
    raise FileNotFoundError(
        "Abre el notebook dentro de la carpeta PRE_11_vuelos."
    )

submission = base / "submission"
submission.mkdir(parents=True, exist_ok=True)

archivos = [
    "carrier_summary.csv",
    "day_hour_delay.csv",
    "monthly_national_kpis.csv",
    "overall_kpis.csv",
    "priority_segments.csv",
    "seasonality.csv",
]

for nombre in archivos:
    ruta = submission / nombre

    # Conservar cualquier archivo que ya exista.
    if not ruta.exists():
        ruta.write_text("status\n", encoding="utf-8")

    assert ruta.is_file()
    print(f"OK: {nombre}")

print(f"\nLos 6 archivos existen en: {submission}")
```
